```python
import scipp as sc
from drcspec.instrument import Instrument


T_OFFSET = sc.scalar(1.7, unit="ms")
central_wavelength = sc.scalar(1, unit="Å")

cspec = Instrument(
    wavelength=central_wavelength,
    f_M=sc.scalar(156, unit="Hz"),
    n=2,
    phase_bw3=sc.scalar(0, unit="deg"),
    t_offset=T_OFFSET,
)
print(cspec)
```
